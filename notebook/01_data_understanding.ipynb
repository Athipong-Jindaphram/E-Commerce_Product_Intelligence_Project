```python
from sqlalchemy import create_engine

username = "postgres"
password = "root"
host = "localhost"
port = "5432"
database = "ecommerce_analyze"

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

tables_to_load = {
    "users": df_users,
    "sessions": df_sessions,
    "reviews": df_reviews,
    "purchases": df_purchases,
    "products": df_products,
    "interactions" : df_interactions
}

for table_name, df_data in tables_to_load.items():
    df_data.to_sql(table_name, engine, if_exists="replace", index=False)
    print(f"Data Successfully Loaded into table '{table_name}' in database '{database}'.")

print("--- All 5 tables have been loaded successfully! ---")
```
